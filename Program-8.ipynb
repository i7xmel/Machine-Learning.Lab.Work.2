```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python

df = pd.read_csv('/content/ObesityDataSet_raw_and_data_sinthetic.csv')
df.head(5)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.columns
```

```python
for k in range(1,6):
  kmeans=KMeans(n_clusters=k)
  kmeans.fit(df[['Age','Weight']])
  df[f'KMeans_{k}'] = kmeans.labels_
  plt.scatter(x=df['Age'], y=df['Weight'], c=df[f'KMeans_{k}'], label=f'K={k}')

plt.xlabel('Age')
plt.ylabel('Weight')
plt.title('K-Means Clustering')
plt.legend()
plt.show()
```

```python
fig, axs = plt.subplots(nrows=1, ncols=5, figsize=(20, 5))

for i, ax in enumerate(fig.axes, start=1):
    ax.scatter(x=df['Age'], y=df['Weight'], c=df[f'KMeans_{i}'])
    ax.set_ylim(0, 200)
    ax.set_xlim(0, 100)
    ax.set_title(f'N Clusters: {i}')

plt.show()
```

# Elbow method

 To determine the optimal number of clusters in k-means clustering. The SSE is a measure of how well the data points fit within their assigned clusters.

```python
sse = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=2)
    kmeans.fit(df[['Age', 'Weight']])
    sse.append(kmeans.inertia_)

plt.plot(range(1, 11), sse, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Sum of Squared Error (SSE)')
plt.show()
```

The optimal number of clusters in the elbow plot is 6. the SSE starts to flatten out at k=6.
