```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import pairwise_distances_argmin_min, silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
num_samples = 750
num_features = 2
max_value = 24
dataset = np.random.rand(num_samples, num_features) * max_value
```

```python
print("Dataset size:", dataset.shape)
print("Sample rows:")
print(dataset[:5])
```

 Apply K-means clustering for clusters ranging from 2 to 10


 Print number of samples in every cluster

  Find within-cluster sum of square distance

  Calculate silhouette score

   Visualize the clusters

  Plot within-cluster sum of square distance

   Plot silhouette scores

   Perform hierarchical clustering

   Use three different linkage measures for presenting hierarchical clusters

   Visualize the clusters formed using various linkages using dendrograms

```python

within_cluster_sumsq = []
silhouette_scores = []

for n_clusters in range(2, 11):
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(dataset)
    centers = kmeans.cluster_centers_


    for i in range(n_clusters):
        cluster_samples = np.sum(labels == i)
        print(f'Cluster {i + 1}: {cluster_samples} samples')


    distances = pairwise_distances_argmin_min(dataset, centers[labels])
    within_cluster_sum_sq = np.sum(distances[1] ** 2)
    print(f'Within-cluster sum of square distance: {within_cluster_sum_sq}')

    within_cluster_sumsq.append(within_cluster_sum_sq)


    silhouette_avg = silhouette_score(dataset, labels)
    silhouette_scores.append(silhouette_avg)
    print(f'Silhouette Score: {silhouette_avg}')


    plt.figure(figsize=(8, 6))
    plt.scatter(dataset[:, 0], dataset[:, 1], c=labels, cmap='viridis', alpha=0.7, edgecolors='k')
    plt.scatter(centers[:, 0], centers[:, 1], c='red', marker='X', s=200, label='Centroids')
    plt.title(f'K-means Clustering with {n_clusters} Clusters')
    plt.legend()
    plt.show()


plt.figure(figsize=(16, 6))

plt.subplot(1, 2, 1)
plt.plot(range(2, 11), within_cluster_sumsq, marker='o')
plt.title('Within-Cluster Sum of Square Distance vs Number of Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Within-Cluster Sum of Square Distance')


plt.subplot(1, 2, 2)
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.title('Silhouette Score vs Number of Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')

plt.tight_layout()
plt.show()


linkage_methods = ['ward', 'complete', 'average']

for linkage_method in linkage_methods:

    Z = linkage(dataset, method=linkage_method)


    plt.figure(figsize=(12, 8))
    dendrogram(Z)
    plt.title(f'Hierarchical Clustering Dendrogram ({linkage_method} linkage)')
    plt.xlabel('Sample Index')
    plt.ylabel('Distance')
    plt.show()
```

Evaluatinhg quality of clusters formed using Kmeans and hierarchical clustering.

```python
silhouette_scores_hierarchical = []

for method in linkage_methods:
    hierarchical = AgglomerativeClustering(n_clusters=3, linkage=method)
    hierarchical_labels = hierarchical.fit_predict(dataset)
    silhouette_avg_hierarchical = silhouette_score(dataset, hierarchical_labels)
    silhouette_scores_hierarchical.append(silhouette_avg_hierarchical)

plt.figure(figsize=(8, 6))
plt.bar(linkage_methods, silhouette_scores_hierarchical, color='skyblue')
plt.title('Silhouette Score for Hierarchical Clustering')
plt.xlabel('Linkage Method')
plt.ylabel('Silhouette Score')
plt.show()

```

 A silhouette score of 1 indicates perfect separation, while a score of -1 indicates that the clusters are completely overlapping. The silhouette score in my graph is 0.35, which is considered to be a moderate score. This means that the clusters are somewhat separated, but there is still some overlap between them
