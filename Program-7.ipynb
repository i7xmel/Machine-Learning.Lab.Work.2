```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, load_breast_cancer, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
iris = load_iris()
breast_cancer = load_breast_cancer()
diabetes = load_diabetes()

datasets = [iris, breast_cancer, diabetes]
dataset_names = ['Iris', 'Breast Cancer', 'Diabetes']
```

```python
def apply_pca(X, n_components):
    pca = PCA(n_components=n_components)
    X_pca = pca.fit_transform(X)
    variance_ratio = pca.explained_variance_ratio_
    return X_pca, variance_ratio

def apply_lda(X, y, n_components):
    lda = LDA(n_components=n_components)
    X_lda = lda.fit_transform(X, y)
    return X_lda

def train_classifiers(X_train, y_train, X_test):
    classifiers = {
        "KNN": KNeighborsClassifier(),
        "Decision Tree": DecisionTreeClassifier(),
        "Random Forest": RandomForestClassifier()
    }

    trained_classifiers = {}
    for name, clf in classifiers.items():
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        trained_classifiers[name] = y_pred

    return trained_classifiers

def evaluate_classifiers(y_true, predictions):
    metrics = {
        'Accuracy': accuracy_score,
        'Precision': precision_score,
        'Recall': recall_score,
        'F1 Score': f1_score
    }

    results = {}
    for metric_name, metric_func in metrics.items():
        results[metric_name] = {
            name: metric_func(y_true, pred) for name, pred in predictions.items()
        }

    return results
```

```python
for idx, data in enumerate(datasets):
    X = data.data
    y = data.target

    print(f"Dataset: {dataset_names[idx]}")
    print(f"Number of features before dimensionality reduction: {X.shape[1]}")
```

```python
 X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python

# Standardize the features
scaler = StandardScaler()
X_train_std = scaler.fit_transform(X_train)
X_test_std = scaler.transform(X_test)
```
